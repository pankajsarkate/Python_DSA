# 3. Alternating Sign Series: Create a function to solve: 1 - 2 + 3 - 4 + 5 ... up to n terms.

#### INPUT: n = 5
|
#### OUTPUT: 3 (1-2+3-4+5)

```python
n = 5
def alternating_sign_series(n):
    temp = ''
    for i in range(1, n+1):
        temp = temp + str(i)
        if i %2!=0:
            temp = temp + '-'
        else:
            temp = temp + '+'

    indx = []
    for i in temp:
        if i.isdigit():
            indx.append(i)
    num = (len(indx)//2)+1


    print(f"{num}({temp.rstrip('-')})")

alternating_sign_series(n)
```

# 4. Power Series: Write a function power_sum(x, n) to compute: 1 + x + x² + x³ + ... + xⁿ.

#### INPUT: x = 2, n = 3
#### OUTPUT: 15 (1 + 2 + 4 + 8)

```python
x, n = 2, 3

def power_sum(x, n):
    temp = ''
    lst = []
    for i in range(0, n+1):
        temp += str(x**i) + '+'
    for i in temp:
        if i.isdigit():
            lst.append(int(i))
    total = sum(lst)
    print(f"{total}({temp.rstrip('+')})")

power_sum(x, n)
```
